```python
from pathlib import Path
import pandas as pd
import cv2
import matplotlib.pyplot as plt
```

```python
from pathlib import Path

dataset_root = Path("../datasets")

fruits = ["Apple", "Banana", "Tomato"]

for fruit in fruits:

    data_path = dataset_root / fruit / "Training"

    print(f"\n{fruit}")

    variety_count = 0

    for variety in data_path.iterdir():

        if variety.is_dir():

            variety_count += 1

            image_count = len(list(variety.iterdir()))

            print(f"{variety.name}: {image_count} images")

    print(f"Total varieties: {variety_count}")
```

```python
apple_train = dataset_root / "Apple" / "Training"

for variety in apple_train.iterdir():

    if variety.is_dir():

        image_count = len(list(variety.iterdir()))

        print(variety.name, image_count)
```
